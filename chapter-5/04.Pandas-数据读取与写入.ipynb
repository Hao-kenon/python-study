```python
import pandas as pd
#数据读取
df = pd.read_csv("data/sales.csv",usecols=['订单号','产品类别','销售数量','单价'])

#数据处理
df['销售额'] = df['销售数量'] * df['单价']

#数据写入
df.to_csv("data/sales_01.csv",index=False)
```
