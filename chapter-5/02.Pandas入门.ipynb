```python
import pandas as pd

# 构造DataFrame -- 创建数据集(学员成绩信息)  *** data frame 就是一张表
df = pd.DataFrame([
    {"姓名":"王","语文":80,"数学":90,"英语":75},
    {"姓名":"张","语文":75,"数学":85,"英语":90},
    {"姓名":"李","语文":90,"数学":80,"英语":85},
    {"姓名":"赵","语文":85,"数学":95,"英语":80},
    {"姓名":"孙","语文":95,"数学":85,"英语":90},
    {"姓名":"周","语文":65,"数学":75,"英语":80},
    {"姓名":"吴","语文":70,"数学":80,"英语":75},
    {"姓名":"郑","语文":80,"数学":85,"英语":90},
    {"姓名":"王","语文":80,"数学":90,"英语":75},
])
df
# 统计计算
print(f"语文最大值:{df['语文'].max()},语文最小值:{df['语文'].min()},语文平均值:{df['语文'].mean():.2f}")
print(f"数学最大值:{df['数学'].max()},数学最小值:{df['数学'].min()},数学平均值:{df['数学'].mean():.2f}")
print(f"英语最大值:{df['英语'].max()},英语最小值:{df['英语'].min()},英语平均值:{df['英语'].mean():.2f}")
```
