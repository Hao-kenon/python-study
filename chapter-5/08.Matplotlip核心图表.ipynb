```python
import matplotlib.pyplot as plt
from matplotlib.axes import Axes

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei'] #设置中文字体

#创建子图
# figure :画布对象 axes:子图数组
figure,axes = plt.subplots(nrows = 1,ncols = 2,figsize=(20,6),dpi = 100) #nrow:行 ncols:列

# 图一 柱状图
countries1 = ['美国','中国','英国','日本','德国']
oil_reserves = [30,100,150,75,110]

axes1: Axes = axes[0]
axes1.bar(countries1,oil_reserves,color='g',width=0.4)
axes1.set_title('世界石油储备',fontsize=18)
axes1.set_xlabel('国家',fontsize=14)
axes1.set_ylabel('石油储备(亿桶)',fontsize=14)
axes1.grid(linestyle='--',alpha=0.6)

# 图二 饼状图
countries2 = ['美国','中国','英国','日本','德国']
populations = [3.4,14.09,0.64,1.26,0.83]

axes2:Axes = axes[1]
axes2.set_title('世界人口占比',fontsize=18)
axes2.pie(populations,labels=countries2,autopct='%1.1f%%') #饼状图
axes2.legend(loc='lower center',ncol=5,bbox_to_anchor=(0.5,-0.05)) # ncol显示图例每行个数，bbox_to_anchor设置图例位置(控制左右，控制上下)

#保存图片
plt.savefig('data/01.png')
plt.show()
```
